```python
# ============================================================
# RANDOM FOREST CLASSIFICATION - WINE QUALITY
# Complete code in ONE Google Colab cell
# ============================================================

# -------------------- 1. IMPORT LIBRARIES --------------------

import pandas as pd
import matplotlib.pyplot as plt

# Used to split data into training and testing sets
from sklearn.model_selection import train_test_split

# Used to convert "bad" and "good" into 0 and 1
from sklearn.preprocessing import LabelEncoder

# Random Forest classification model
from sklearn.ensemble import RandomForestClassifier

# Used to check model performance
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

print("Libraries imported successfully!")


# -------------------- 2. UPLOAD DATASET --------------------

# This allows you to upload winequality.csv from your computer
from google.colab import files

uploaded = files.upload()

# Read the CSV file into a pandas DataFrame
wine = pd.read_csv("winequality.csv")

print("\nDataset loaded successfully!")


# -------------------- 3. VIEW THE DATA --------------------

# Display first 5 rows
print("\nFirst 5 rows:")
display(wine.head())

# Display number of rows and columns
print("\nDataset shape:")
print(wine.shape)

# Display information about columns and data types
print("\nDataset information:")
wine.info()

# Display statistical summary
print("\nStatistical summary:")
display(wine.describe())

# Check column names
print("\nColumn names:")
print(wine.columns)


# -------------------- 4. CHECK MISSING VALUES --------------------

# isnull() checks for missing values
# sum() counts the missing values in each column
print("\nMissing values:")
print(wine.isnull().sum())


# -------------------- 5. CONVERT QUALITY INTO 2 CLASSES --------------------

# Original quality values are converted into:
# 2 to 6.5  -> bad
# 6.5 to 8  -> good

bins = (2, 6.5, 8)
group_names = ["bad", "good"]

wine["quality"] = pd.cut(
    wine["quality"],
    bins=bins,
    labels=group_names
)

print("\nQuality after conversion:")
print(wine["quality"].value_counts())


# -------------------- 6. CONVERT TEXT INTO NUMBERS --------------------

# Machine Learning models work with numbers.
# LabelEncoder changes:
# bad  -> 0
# good -> 1

encoder = LabelEncoder()

wine["quality"] = encoder.fit_transform(wine["quality"])

print("\nQuality after Label Encoding:")
print(wine["quality"].value_counts())


# -------------------- 7. DISPLAY QUALITY GRAPH --------------------

wine["quality"].value_counts().plot(kind="bar")

plt.xlabel("Wine Quality")
plt.ylabel("Number of Wines")
plt.title("Wine Quality Distribution")
plt.show()


# -------------------- 8. SEPARATE INPUT AND OUTPUT --------------------

# X contains the input features used to make predictions
# quality is removed because it is our target/output

X = wine.drop("quality", axis=1)

# y contains the target/output we want to predict
y = wine["quality"]

print("\nInput features (X):")
print(X.columns)

print("\nTarget (y):")
print(y.name)


# -------------------- 9. SPLIT DATA --------------------

# 80% data is used for training
# 20% data is used for testing
# random_state=42 makes the same split every time

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))


# -------------------- 10. CREATE RANDOM FOREST MODEL --------------------

# Random Forest creates many Decision Trees
# n_estimators=100 means 100 trees will be created
# random_state=0 makes the model results reproducible

model = RandomForestClassifier(
    n_estimators=10,
    random_state=0
)

print("\nRandom Forest model created!")


# -------------------- 11. TRAIN THE MODEL --------------------

# fit() teaches the Random Forest using training data

model.fit(X_train, y_train)

print("Model trained successfully!")


# -------------------- 12. MAKE PREDICTIONS --------------------

# predict() uses the trained model to predict the quality
# of wines in the testing data

y_pred = model.predict(X_test)

print("\nPredictions made successfully!")


# -------------------- 13. SHOW ACTUAL VS PREDICTED --------------------

comparison = pd.DataFrame({
    "Actual Quality": y_test.values,
    "Predicted Quality": y_pred
})

print("\nActual vs Predicted:")
display(comparison.head(20))


# -------------------- 14. CHECK ACCURACY --------------------

# Accuracy tells us how many predictions were correct

accuracy = accuracy_score(y_test, y_pred)

print("\nRandom Forest Accuracy:")
print(accuracy)

print("\nAccuracy in percentage:")
print(f"{accuracy * 100:.2f}%")


# -------------------- 15. CLASSIFICATION REPORT --------------------

# Shows:
# Precision = how many predicted cases were actually correct
# Recall    = how many actual cases were correctly found
# F1-score  = balance between precision and recall

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=encoder.classes_
    )
)


# -------------------- 16. CONFUSION MATRIX --------------------

# Confusion matrix shows:
# Correct and incorrect predictions for each class

cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(cm)

# Display the confusion matrix as a graph
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=encoder.classes_
)

disp.plot()

plt.title("Random Forest Confusion Matrix")
plt.show()


# -------------------- 17. PREDICT A NEW WINE --------------------

# Enter values for one new wine.
# The values must be in the same order as X.columns.

new_wine = pd.DataFrame([[
    7.4,    # fixed acidity
    0.70,   # volatile acidity
    0.00,   # citric acid
    1.9,    # residual sugar
    0.076,  # chlorides
    11.0,   # free sulfur dioxide
    34.0,   # total sulfur dioxide
    0.9978, # density
    3.51,   # pH
    0.56,   # sulphates
    9.4     # alcohol
]], columns=X.columns)

# Use the trained Random Forest to predict the new wine
prediction = model.predict(new_wine)

# Convert 0/1 back into "bad"/"good"
prediction_label = encoder.inverse_transform(prediction)

print("\nNew Wine Prediction:")
print("Predicted Wine Quality:", prediction_label[0])


# ============================================================
# END OF RANDOM FOREST PROGRAM
# ============================================================
```
